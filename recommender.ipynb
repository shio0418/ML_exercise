```python
from surprise import accuracy, Dataset, SVD
from surprise import Dataset
from surprise.model_selection import cross_validate
from surprise.model_selection import train_test_split

# 図示のため以下をインポート
import matplotlib.pyplot as plt

# Load the movielens-100k dataset (download it if needed).
data = Dataset.load_builtin('ml-100k')

# sample random trainset and testset
# test set is made of 25% of the ratings.
#trainset, testset = train_test_split(data, test_size=0.25)
trainset, testset = train_test_split(data)

# 2. biasedを変えてみる
rmse_means_T = []
rmse_means_F = []
factors = range(1,21,1)

for n in factors:
    # TRUE のとき
    # Train the algorithm on the trainset, and predict ratings for the testset
    algo = SVD(n_factors=n, biased=True, reg_all=0.0001, random_state=0)

    # kfold cross validationの実施
    results = cross_validate(algo, data, measures=['RMSE'], cv=5)

    # Then compute RMSE
    mean_rmse = results['test_rmse'].mean()
    rmse_means_T.append(mean_rmse)

    print(f"biased={True}, mean RMSE={mean_rmse:.4f}")

    # ------------------ #

    # FALSE のとき
    # Train the algorithm on the trainset, and predict ratings for the testset
    algo = SVD(n_factors=n, biased=False, reg_all=0.0001, random_state=0)

    # kfold cross validationの実施
    results = cross_validate(algo, data, measures=['RMSE'], cv=5)

    # Then compute RMSE
    mean_rmse = results['test_rmse'].mean()
    rmse_means_F.append(mean_rmse)

    print(f"biased={False}, mean RMSE={mean_rmse:.4f}")

# True,FalseのRMSEをそれぞれ図示
plt.plot(factors, rmse_means_T,color='red',label='biased=True')
plt.plot(factors, rmse_means_F,color='blue',label='biased=False')
plt.xlabel('n_factors')
plt.ylabel('Mean RMSE')
plt.legend()
plt.show()


# biased項があったほうがRMSEが減少してる
```

```python
# 3. reg all(lambda) を変えてみる
reg_alls = [1e-7,5e-7,1e-6,5e-6,1e-5,5e-5,1e-4,5e-4,1e-3,5e-3,1e-2,5e-2,1e-1,5e-1]
rmse_means_r = []

for r in reg_alls:
    # Train the algorithm on the trainset, and predict ratings for the testset
    algo = SVD(n_factors=6, biased=True, reg_all=r, random_state=0)
    #algo.fit(trainset)

    # kfold cross validationの実施
    results = cross_validate(algo, data, measures=['RMSE'], cv=5)

    # Then compute RMSE
    mean_rmse = results['test_rmse'].mean()
    rmse_means_r.append(mean_rmse)

    print(f"reg_all={r}, mean RMSE={mean_rmse:.4f}")

# RMSEをグラフで図示
plt.plot(reg_alls, rmse_means_r)
plt.xscale('log')
plt.xlabel('reg_all')
plt.ylabel('Mean RMSE')
plt.legend()
plt.show()
```

```python
# 4. 最高精度を達成

# 今までの結果を踏まえ、それぞれのパラーメータで精度がよかった範囲をループで探索
# biased項は、先程の結果よりすべての場合においてTrueとしたほうが精度が高いと仮定

min_rmse = float('inf')
best_parameters = []
rmse_means = []
factors = range(3,20,1)
reg_alls = [1e-7,5e-7,1e-6,5e-6,1e-5,5e-5,1e-4,5e-4,1e-3,5e-3,1e-2,5e-2]
#reg_alls = np.logspace(-5,-2,20)

for n in factors:
  for r in reg_alls:
    # Train the algorithm on the trainset, and predict ratings for the testset
    algo = SVD(n_factors=n, biased=True, reg_all=0.0001, random_state=0)
    #algo.fit(trainset)

    # kfold cross validationの実施
    results = cross_validate(algo, data, measures=['RMSE'], cv=5)

    # Then compute RMSE
    #predictions = algo.test(testset)
    #print(f'---n_factors = {n}---')
    #rmse = accuracy.rmse(predictions)
    mean_rmse = results['test_rmse'].mean()
    rmse_means.append(mean_rmse)

    print(f"n_factors={n}, reg_alls={r}, mean RMSE={mean_rmse:.4f}")

    if min_rmse > mean_rmse:
      min_rmse = mean_rmse
      best_parameters = [n,r]
print(f"best n_factors={best_parameters[0]}, best feg_alls={best_parameters[1]}, best RMSE={min_rmse:.4f}")


# RMSEをグラフで図示
#plt.plot(factors, rmse_means_n)
#plt.xlabel('n_factors')
#plt.ylabel('Mean RMSE')
#plt.legend()
#plt.show()
```
